```python
import os
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Parameters
N = 50                     # number of users
weeks = range(1, 9)        # 8-week program
rows = []

for u in range(1, N + 1):
    age = np.random.randint(20, 58)
    sex = np.random.choice(["F", "M"], p=[0.6, 0.4])
    height = np.random.randint(155, 185)
    start_w = np.random.uniform(60, 100)
    target = start_w - 10
    act = np.random.choice(
        ["sedentary", "light", "moderate", "active"],
        p=[0.25, 0.35, 0.3, 0.1]
    )

    w = start_w
    for wk in weeks:
        # Daily intake around TDEE – simulate deficit variability
        intake = np.random.uniform(1600, 2400)
        workout = np.random.randint(40, 300)

        # Simulate weight change, avg ~0.35 kg loss per week
        delta = np.random.normal(-0.35, 0.25)

        # Occasional plateau
        if np.random.rand() < 0.15:
            delta = np.random.normal(-0.05, 0.08)

        new_w = max(40, w + delta)
        rows.append({
            "user_id": u,
            "age": age,
            "sex": sex,
            "height": height,
            "weight": round(new_w, 1),
            "target_weight": round(target, 1),
            "activity_level": act,
            "calorie_intake": int(intake),
            "workout_mins": int(workout),
            "week_no": wk,
            "weight_change": round(new_w - w, 2)
        })
        w = new_w

# Create DataFrame
df = pd.DataFrame(rows)

# Save safely — use /tmp so it always works, even in read-only dirs
save_path = "/tmp/patients.csv"
df.to_csv(save_path, index=False)
print(f"Wrote dataset to {save_path}")

# Preview
df.head()
```
